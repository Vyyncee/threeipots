```python
def find_leaky_columns(df, label_col="label", ratio_threshold=10, constant_threshold=1):
    """
    Détecte les colonnes fuites dans un dataset binaire (Clean / Infected).

    - ratio_threshold : si (unique_clean / unique_infected) > ratio_threshold → fuite
    - constant_threshold : si une classe a <= constant_threshold valeurs → fuite
    """

    cols_to_remove = []
    clean_df = df[df[label_col] == "Clean"]
    infected_df = df[df[label_col] == "Infected"]

    for col in df.columns:
        if col == label_col:
            continue

        unique_clean = clean_df[col].nunique()
        unique_infected = infected_df[col].nunique()

        # règle 1 : colonne quasi constante dans une classe
        if unique_clean <= constant_threshold or unique_infected <= constant_threshold:
            cols_to_remove.append(col)
            continue

        # règle 2 : asymétrie énorme
        ratio = max(unique_clean, unique_infected) / max(1, min(unique_clean, unique_infected))
        if ratio >= ratio_threshold:
            cols_to_remove.append(col)
            continue

        # règle 3 : colonnes identifiantes
        if any(key in col.lower() for key in ["addr", "ip.", "eth.", "port", "checksum", "id", "seq", "stream", "flags"]):
            cols_to_remove.append(col)
            continue

    return sorted(set(cols_to_remove))
```

```python
# Supprimer les colonnes qui leak normal/attaque

for key in merged_datasets.keys():
    datas.drop_leaky_columns(key, find_leaky_columns(datas.merged[key]))
```

```python
# Déclaration des model et des hyper paramètre a tester

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

models = {
    'logreg': {
        'model': LogisticRegression(),
        'params': {
            'model__C': [0.01, 0.1, 1, 10, 100],
            'model__solver': ['liblinear']
        }
    },

    'rf': {
        'model': RandomForestClassifier(),
        'params': {
            'model__n_estimators': [50, 100, 200, 500],
            'model__max_depth': [None, 5, 10, 20, 50],
            'model__min_samples_split': [2, 5, 10, 20],
            'model__min_samples_leaf': [1, 2, 4, 8],
            'model__max_features': ['sqrt', 'log2', None]
        }
    },

    'xgb': {
        'model': XGBClassifier(eval_metric='logloss'),
        'params': {
            'model__n_estimators': [50, 100, 200, 500],
            'model__max_depth': [3, 5, 7, 10],
            'model__learning_rate': [0.001, 0.01, 0.05, 0.1, 0.2],
            'model__subsample': [0.5, 0.7, 1.0],
            'model__colsample_bytree': [0.5, 0.7, 1.0],
            'model__gamma': [0, 0.1, 0.2, 0.5]
        }
    }
}
```

# GridSearch / Entrainement et Ensemble Learning (Voting)

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import VotingClassifier

pipelines_by_protocol =  {} # Pour stocker les pipelines par protocole

for key, df in balanced.items(): # Pour chaque protocole

    print('GridSearch pour ' + key)
    preprocessor = DataManager.get_preprocessor(df) # Récupération du preprocessor

    X_train, X_test, y_train, y_test = DataManager.mix_split(df, 3000) # Mix and Split for GrisdSearch (Il n'y aura que 80% de 3000 dans le train)

    best_pipelines = {} # Pour sauvegarder les models 

    for name, value in models.items(): # Pour chaque model et param
        pipeline = Pipeline([('preprocessor', preprocessor), ('model', value['model'])]) # Création de pipeline

        grid = GridSearchCV(pipeline, value['params'], cv=5, scoring='accuracy', n_jobs=-1) # Le grid search pour chercher les meilleurs paramètres
        grid.fit(X_train, y_train) # On envoie le grid search sur le bon dataset de train

        # On sauvegarde la meilleure pipeline (Celle qui a le model avec les meilleures params)
        best_pipelines[name] = grid.best_estimator_

        print(f"{name}: score={grid.best_score_:.4f}, params={grid.best_params_}")
    
    # Entrainement et Evaluation sur le dataset a 80 %
    X_train, X_test, y_train, y_test = DataManager.mix_split(df)

    for name, pipeline in best_pipelines.items():
        y_pred = pipeline.predict(X_test)
        # Evaluation
        ModelManager.eval(y_pred, y_test)

    # VOTING
    estimators = [(name, pipe) for name, pipe in best_pipelines.items()]
    voting = VotingClassifier(
        estimators=estimators,
        voting='hard' # Vote majoritaire (Comme 3 model c'est ok)
    )
    voting.fit(X_train, y_train)

    # Eval
    y_pred = voting.predict(X_test)
    ModelManager.eval(y_pred, y_test)
    
    # Sauvegarde
    ModelManager.save(voting, key + ModelManager.SUFFIX_FOR_SAVE_VOTING)
```
